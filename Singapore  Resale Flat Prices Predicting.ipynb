```python
def convert_year(year):
    y = year.split()
    if len(y) == 4:
        total = round(int(y[0]) + (int(y[2]) / 12), 1)
    else:
        total = round(int(y[0]))
    return total
```

## ML model selection

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from statistics import mean
```

```python
df_numeric = df_merge.select_dtypes(include=['number'])
df_numeric.head()
```

```python
df_numeric.columns
```

```python
target_col = 'resale_price'

X_df = df_numeric.drop(columns=[target_col])
y_df = df_numeric[target_col]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_df, y_df, test_size=0.2, random_state=42)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
models = {
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=20, random_state=42, n_jobs=-1),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=6,
                            random_state=42, n_jobs=-1, verbosity=0)
}
```

```python
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)

    results[name] = {"R2": r2, "RMSE": rmse, "MAE": mae}
```

```python
results
```

```python
results_df = pd.DataFrame(results).T
results_df
```

```python
# Do we need to fine-tune the model parameters?

# Not strictly necessary:
# - Our Random Forest model already gives very strong performance 
#   (R² ~0.98, low RMSE and MAE).
# - This means the model captures the patterns in resale price quite well 
#   without tuning.

# Why fine-tuning can still help:
# - It might squeeze out a bit more accuracy (slightly lower errors).
# - It demonstrates knowledge of model optimization during presentations/interviews.
# - But the gain is usually marginal compared to the baseline Random Forest.

# Conclusion:
# - Fine-tuning is not mandatory since the baseline is already very strong.
# - If time permits, we can do light tuning (RandomizedSearchCV).
# - Otherwise, Random Forest with default parameters is already a reliable choice.
```

```python
rf_model = RandomForestRegressor(n_estimators=20, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)
```

```python
# Example user input (already encoded & cleaned like your training data)
user_input = [[2015, 80, 30, 12, 3, 5, 7, 2]]  
# ['year', 'floor_area_sqm', 'remaining_lease',
# 'town_encoded', 'flat_type_encoded', 'street_name_encoded',
# 'storey_range_encoded', 'flat_model_encoded']

# Predict resale price
predicted_price = rf_model.predict(user_input)
print(predicted_price)
```
